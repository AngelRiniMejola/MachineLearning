```python
#Pandas has all functions related to database tables
import pandas as pd
```

```python
#Assigning the dataset to a vriable
dataset=pd.read_csv("Salary_Data.csv")
```

```python
#printing the dataset
dataset
```

```python
#splitting input and assigning to a variable -independent
independent=dataset[["YearsExperience"]]
```

```python
#printing the splitted input dataset
independent
```

```python
#splitting output  and assigning to a variable -independent
dependent=dataset[["Salary"]]
```

```python
#printing the splitted output dataset
dependent
```

```python
#splitting training set and test set
```

```python
from sklearn.model_selection import train_test_split
```

```python
#x refers input, y refers output
x_train,x_test,y_train,y_test = train_test_split(independent,dependent,test_size=0.30,random_state=0)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
regressor=LinearRegression()
```

```python
#below line will create model
regressor.fit(x_train,y_train)
```

```python
LinearRegression()
```

```python
weight=regressor.coef_
```

```python
weight
```

```python
bias=regressor.intercept_
```

```python
bias
```

```python
#storing predicted output in y_pred 
y_pred=regressor.predict(x_test)
```

```python
y_pred
```

```python
#evaluating the predicted output
from sklearn.metrics import r2_score
```

```python
#r_score has value of actual output - predicted output. If r_score has value nearly 1, it indicates the created model 
# has better performance else the model has poor performance
r_score=r2_score(y_test,y_pred)
```

```python
r_score
```

```python
#To save the best model
import pickle
filename="Finalized_linear_model.sav"
pickle.dump(regressor,open(filename,'wb'))
```

```python
#To load the saved model
load_model=pickle.load(open("Finalized_linear_model.sav",'rb'))
```

```python
#the predicted output value given for input will be stored in result variable
result=load_model.predict([[0]])
```
